```python
# PyTorch 텐서 생성과 연산에 필요한 라이브러리를 불러온다.
import torch
```

```python
# 0부터 11까지의 값을 가진 "1차원 float32 텐서"를 만든다.
# 핵심: arange(12)는 0 이상 12 미만의 값을 만들며, numel은 전체 원소 수이고 shape는 축별 길이다.
x = torch.arange(12, dtype=torch.float32)

print(x)
print(x.numel()) # 원소의 개수
print(x.shape)
```

```python
# reshape는 원소의 개수와 순서를 유지하면서 텐서의 모양만 바꾼다.
# 핵심: reshape는 원소 수와 순서를 보존하며, 가능한 경우 원본과 같은 저장 공간을 공유한다.
X = x.reshape(3, 4)

print(X)
print(X.shape)
print(x[3])
print(X[0, 3])
print(x[3] == X[0, 3])
```

```python
# -1로 지정한 차원의 크기는 전체 원소 수에 맞춰 자동으로 계산된다.
# 핵심: -1은 한 축에만 사용할 수 있고 전체 원소 수에 맞춰 그 축의 길이를 자동 계산한다.
X1 = x.reshape(-1, 4)
X2 = x.reshape(3, -1)

print(X1)
print(X2)
print(X1.shape)
print(X2.shape)
```

```python
# zeros, ones, randn은 원하는 shape의 텐서를 빠르게 초기화한다.
# 핵심: zeros와 ones는 상수 초기화, randn은 평균 0·표준편차 1인 정규분포 표본을 생성한다.
zeros = torch.zeros((2, 3, 4))
ones = torch.ones((2, 3, 4))
random_values = torch.randn(3, 4)

print(zeros)
print(zeros.shape)

print(ones)
print(ones.shape)

print(random_values)
print(random_values.shape)
```

```python
# torch.tensor는 Python 리스트에 지정한 값으로 텐서를 직접 만든다.
# 핵심: 정수 리스트로 tensor를 만들면 기본 dtype은 int64이므로 모델 입력에서는 dtype을 확인해야 한다.
values = torch.tensor([
    [2, 1, 4, 3],
    [1, 2, 3, 4],
    [4, 3, 2, 1],
])

print(values)
print(values.shape)
print(values.dtype)
```
