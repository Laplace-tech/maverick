```python
# 인덱싱과 슬라이싱을 연습할 3 x 4 텐서를 만든다.
import torch

X = torch.arange(12, dtype=torch.float32).reshape(3, 4)

print("Original X: ")
print(X)
print("X.shape:", X.shape)
```

```python
# 음수 인덱스는 뒤에서부터 접근하며, 슬라이스의 끝 인덱스는 포함하지 않는다.
last_row = X[-1]
selected_rows = X[1:3]

print("\nLast row:")
print(last_row)
print("shape:", last_row.shape)

print("\nRows 1 to 2:")
print(selected_rows)
print("shape:", selected_rows.shape)
```

```python
# 단일 원소 인덱싱은 0차원 텐서를 반환하고, clone은 원본을 보존한다.
single_value = X[1, 2]

print("\nX[1, 2]:")
print(single_value)
print("shape:", single_value.shape)

X_single_modified = X.clone()
X_single_modified[1, 2] = 69

print("\nAfter changing one element:")
print(X_single_modified)

X_slice_modified = X.clone()
X_slice_modified[:2, :] = 12

print("\nAfter changing multiple elements:")
print(X_slice_modified)
```

```python
# 콜론(:)은 해당 차원의 모든 원소를 선택한다.
first_column = X[:, 0]
last_column = X[:, -1]
submatrix = X[:2, 1:3]

print("\nFirst column:")
print(first_column)
print("Shape:", first_column.shape)

print("\nLast column:")
print(last_column)
print("Shape:", last_column.shape)

print("\nSubmatrix:")
print(submatrix)
print("Shape:", submatrix.shape)
```
