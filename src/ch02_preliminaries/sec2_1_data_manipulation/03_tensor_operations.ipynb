```python
# PyTorch의 원소별 연산과 텐서 결합 연산을 사용한다.
import torch
```

```python
# torch.exp는 각 원소에 지수 함수를 적용하므로 입력과 출력 shape가 같다.
# 핵심: exp는 각 원소에 독립적으로 적용되는 원소별 연산이므로 입력과 출력 shape가 같다.
x = torch.arange(12, dtype=torch.float32)
x = x.reshape(3, 4)
exp_x = torch.exp(x)

print("Original x:")
print(x)
print("shape:", x.shape)

print("\nElementwise exponential:")
print(exp_x)
print("shape:", exp_x.shape)
```

```python
# 크기가 같은 텐서끼리 산술 연산을 하면 대응하는 원소별로 계산된다.
# 핵심: 같은 shape의 텐서 산술은 대응 원소끼리 계산하며, shape가 다르면 broadcasting 가능성을 검사한다.
x = torch.tensor([1.0, 2, 4, 8])
y = torch.tensor([2.0, 2, 2, 2])

addition = x + y
subtraction = x - y
multiplication = x * y
division = x / y
exponentiation = x ** y

print("Original vectors:")
print("x:", x)
print("y:", y)

print("\nElementwise arithmetic:")
print("x + y:", addition)
print("x - y:", subtraction)
print("x * y:", multiplication)
print("x / y:", division)
print("x ** y:", exponentiation)
```

```python
# torch.cat은 기존 차원을 따라 텐서를 이어 붙이며 dim에 따라 결과 shape가 달라진다.
# 핵심: cat은 새 축을 만들지 않고 지정한 dim의 길이를 더하며 나머지 축의 길이는 같아야 한다.
X = torch.arange(12, dtype=torch.float32).reshape(3, 4)
Y = torch.tensor([
    [2.0, 1, 4, 3],
    [1, 2, 3, 4],
    [4, 3, 2, 1],
])

rows_concatenated = torch.cat((X, Y), dim=0)
columns_concatenated = torch.cat((X, Y), dim=1)

print("Original tensors:")
print("X:")
print(X)
print("Y:")
print(Y)

print("\nConcatenate along axis 0:")
print(rows_concatenated)
print("Shape:", rows_concatenated.shape)

print("\nConcatenate along axis 1:")
print(columns_concatenated)
print("Shape:", columns_concatenated.shape)
```

```python
# 비교 연산은 원소별 True/False를 담은 bool 텐서를 반환한다.
# 핵심: 비교 결과는 bool 텐서이고, bool의 합은 True를 1로 보아 조건을 만족한 원소 수를 센다.
comparison = (X == Y)

print("Elementwise comparison:")
print(comparison)
print("shape:", comparison.shape)
print("dtype:", comparison.dtype)
print("Equal elements:", comparison.sum().item())
```

```python
# sum은 모든 원소를 하나의 0차원 텐서로 축약하고 item은 Python 숫자를 꺼낸다.
# 핵심: 모든 축을 sum하면 0차원 텐서가 되고 item은 이를 autograd 밖의 Python 숫자로 꺼낸다.
total = X.sum()

print("Sum of all elements:")
print(total)
print("shape:", total.shape)
print("python scalar:", total.item())
```
