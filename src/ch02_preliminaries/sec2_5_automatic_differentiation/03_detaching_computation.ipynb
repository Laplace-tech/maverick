```python
import torch
```

```python
# gradient를 추적할 입력 벡터를 생성
x = torch.arange(
    4.0,
    requires_grad=True,
)

# y는 x로부터 만들어졌으므로 계산 그래프에 연결되어 있다.
y = x * x

# u는 y와 같은 값을 가지지만 이전 계산 그래프와 분리된다.
u = y.detach()

print("y:", y)
print("u:", u)
print("Same values:", torch.equal(y, u))
print("y requires_grad:", y.requires_grad)
print("u requires_grad:", u.requires_grad)
print("y grad_fn:", y.grad_fn)
print("u grad_fn:", u.grad_fn)
```

```python
# z = u * x 에서 u는 계산 그래프와 분리되었으므로 상수로 취급
z = u * x

z.sum().backward()

print("x.grad:", x.grad)
print("Expected gradient:", u)

# dz/dx=u이므로 x.grad는 u와 같아야 한다.
assert torch.equal(x.grad, u)
```

```python
# 앞에서 계산한 gradient가 누적되지 않도록 초기화
x.grad.zero_()

# y=x²의 계산 그래프는 detach 이후에도 그대로 존재
y.sum().backward()

expected_gradient = 2 * x.detach()

print("Gradient through y:", x.grad)
print("Expected gradient:", expected_gradient)

assert torch.equal(
    x.grad,
    expected_gradient,
)
```
