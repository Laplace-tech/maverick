# 2.5 Automatic Differentiation Cumulative Test

`...` 부분에 핵심 코드를 채운 뒤 위에서부터 실행하세요. 범위는 2.5.1–2.5.3입니다.

````python
import torch
````

````python
# 문제 1. 연산 기록이 활성화된 벡터 x를 생성하세요.
x = torch.arange(4.0, requires_grad=True)

assert x.shape == (4,)
assert x.requires_grad
assert x.grad is None
````

````python
# 문제 2. y=2xᵀx를 계산하고 backward를 실행하세요.
y = 2 * torch.dot(x, x)
y.backward()

assert y.shape == ()
assert torch.equal(x.grad, 4 * x.detach())
````

````python
# 문제 3. 기존 gradient를 0으로 초기화하세요.
x.grad.zero_()

assert torch.equal(x.grad, torch.zeros_like(x))
````

````python
# 문제 4. 벡터 y=x²에 모든 원소가 1인 upstream gradient를 전달하세요.
y = x * x
upstream_gradient = torch.ones_like(y)
y.backward(gradient=upstream_gradient)

assert upstream_gradient.shape == y.shape
assert torch.equal(x.grad, 2 * x.detach())
````

````python
# 문제 5. 서로 다른 upstream gradient [1,2,3,4]를 전달하세요.
x.grad.zero_()
y = x * x
weights = torch.tensor([1.0, 2.0, 3.0, 4.0])
y.backward(gradient=weights)

expected_gradient = weights * 2 * x.detach()

assert torch.equal(x.grad, expected_gradient)
assert torch.equal(x.grad, torch.tensor([0.0, 4.0, 12.0, 24.0]))
````

````python
# 문제 6. y=x²를 계산 그래프에서 분리한 u를 만드세요.
x.grad.zero_()
y = x * x
u = y.detach()

assert torch.equal(u, y)
assert not u.requires_grad
assert u.grad_fn is None
````

````python
# 문제 7. z=u*x를 미분해 detach된 u가 상수로 취급되는지 확인하세요.
z = u * x
z.sum().backward()

assert torch.equal(x.grad, u)
assert torch.equal(x.grad, x.detach() ** 2)
````
