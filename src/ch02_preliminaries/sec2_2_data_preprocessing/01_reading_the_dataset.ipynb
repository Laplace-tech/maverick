```python
from pathlib import Path

import pandas as pd
```

```python
# VS code가 프로젝트 루트에서 실행되므로 프로젝트의 data 폴더를 사용한다.
data_dir = Path("data")
data_dir.mkdir(exist_ok=True)

data_file = data_dir / "house_tiny.csv"

csv_data = """NumRooms,RoofType,Price
NA,NA,127500
2,NA,106000
4,Slate,178100
NA,NA,140000
"""

data_file.write_text(csv_data, encoding="utf-8")

print("Created:", data_file.resolve())
```

```python
# read_csv.는 CSV 데이터를 행과 열로 구성된 DataFrame으로 읽는다.
data = pd.read_csv(data_file)

print("House dataset:")
print(data)

print("\nshape:", data.shape)
print("Columns:", data.columns.tolist())
```

```python
# CSV의 NA는 pandas에서 결측값 NaN으로 처리된다.
print("Data types:")
print(data.dtypes)

print("\nMissing values per column:")
print(data.isna().sum())

assert data.shape == (4, 3)
assert data["NumRooms"].isna().sum() == 2
assert data["RoofType"].isna().sum() == 3
```
