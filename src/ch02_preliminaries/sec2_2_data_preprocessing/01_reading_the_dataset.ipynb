```python
# 핵심: pandas DataFrame은 행을 사례, 열을 변수로 표현하며 이후 수치 텐서로 변환할 표 형식 데이터다.
from pathlib import Path

import pandas as pd
```

```python
# VS code가 프로젝트 루트에서 실행되므로 프로젝트의 data 폴더를 사용한다.
# 핵심: CSV의 한 행은 하나의 레코드이고 열 이름은 모델의 입력 특징과 목표값을 구분하는 기준이 된다.
data_dir = Path("data")
data_dir.mkdir(exist_ok=True)

data_file = data_dir / "house_tiny.csv"

csv_data = """NumRooms,RoofType,Price
NA,NA,127500
2,NA,106000
4,Slate,178100
NA,NA,140000
"""

data_file.write_text(csv_data, encoding="utf-8")

print("Created:", data_file.resolve())
```

```python
# read_csv.는 CSV 데이터를 행과 열로 구성된 DataFrame으로 읽는다.
# 핵심: read_csv는 CSV를 DataFrame으로 파싱하고 기본 설정에서 NA 같은 표기를 결측값 NaN으로 해석한다.
data = pd.read_csv(data_file)

print("House dataset:")
print(data)

print("\nshape:", data.shape)
print("Columns:", data.columns.tolist())
```

```python
# CSV의 NA는 pandas에서 결측값 NaN으로 처리된다.
# 핵심: NaN이 섞인 수치 열은 부동소수점 dtype이 되며 isna().sum()으로 열별 결측 개수를 확인한다.
print("Data types:")
print(data.dtypes)

print("\nMissing values per column:")
print(data.isna().sum())

assert data.shape == (4, 3)
assert data["NumRooms"].isna().sum() == 2
assert data["RoofType"].isna().sum() == 3
```
