```python
import torch
```

```python
# clone은 값이 같은 독립적인 텐서를 새 메모리에 만든다.
# 핵심: 원소별 덧셈은 shape를 보존하고 clone은 값은 같지만 저장 공간이 독립적인 텐서를 만든다.
A = torch.arange(6, dtype=torch.float32).reshape(2, 3)
B = A.clone()

addition = A + B

print("A:")
print(A)

print("\nB:")
print(B)

print("\nA + B:")
print(addition)
print("Shape:", addition.shape)

assert torch.equal(A, B)
assert A.data_ptr() != B.data_ptr()
assert addition.shape == A.shape
```

```python
# Hadamard product는 같은 위치의 원소끼리 곱한다.
# 핵심: Hadamard product A * B는 같은 위치의 원소끼리 곱하며 일반 행렬 곱과 다르다.
hadamard_product = A * B

print("A * B:")
print(hadamard_product)
print("shape:", hadamard_product.shape)

expected = torch.tensor([
    [0.0, 1.0, 4.0],
    [9.0, 16.0, 25.0],
])

assert torch.equal(hadamard_product, expected)
assert hadamard_product.shape == A.shape
```

```python
# 스칼라는 모든 원소에 자동으로 broadcasting된다.
# 핵심: 스칼라는 텐서의 모든 위치로 broadcasting되므로 연산 전후 tensor shape가 유지된다.
scalar = 2
X = torch.arange(24).reshape(2, 3, 4)

scalar_added = scalar + X
scalar_multiplied = scalar * X

print("Original X:")
print(X)
print("Shape:", X.shape)

print("\n2 + X:")
print(scalar_added)

print("\n2 * X:")
print(scalar_multiplied)
```

```python
# *는 원소별 곱이고 @는 행렬 곱이다.
# 핵심: A * B는 원소별 곱이고 A @ C는 안쪽 차원을 축약하는 행렬 곱이라 결과 shape 규칙이 다르다.
A = torch.arange(6, dtype=torch.float32).reshape(2, 3)
B = torch.ones(2, 3)
C = torch.ones(3, 2)

elementwise_result = A * B
matrix_result = A @ C

print("A * B:")
print(elementwise_result)
print("Shape:", elementwise_result.shape)

print("\nA @ C:")
print(matrix_result)
print("Shape:", matrix_result.shape)
```

```python
# Hadamard product의 gradient도 원소별로 계산된다.
# 핵심: loss = Σ(A ⊙ B)이면 각 Aᵢⱼ에 대한 gradient는 곱해진 상대 원소 Bᵢⱼ다.
A = torch.tensor(
    [[1.0, 2.0], [3.0, 4.0]],
    requires_grad=True,
)
B = torch.tensor([
    [1.0, 9.0],
    [7.0, 4.0],
])

loss = (A * B).sum()
loss.backward()

print("Loss:", loss)
print("Gradient of A:")
print(A.grad)

# loss = Σ(AᵢⱼBᵢⱼ)이므로 ∂loss/∂Aᵢⱼ = Bᵢⱼ다.
assert torch.equal(A.grad, B)
```
