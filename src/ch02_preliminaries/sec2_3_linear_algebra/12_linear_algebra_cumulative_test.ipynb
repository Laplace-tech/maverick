````python
import torch
````

````python
# 문제 1. 0차원 스칼라 텐서 x와 y를 만들고 곱셈을 계산하세요.

x = torch.scalar_tensor(3)
# x = torch.tensor(3)
y = torch.scalar_tensor(4)
# y = torch.tensor(4)

product = x * y

assert x.shape == ()
assert y.shape == ()
assert product.item() == 12.0
````

````python
# 문제 2. 벡터의 길이와 원소 합을 구하세요.
x = torch.tensor([1.0, 2.0, 3.0, 4.0])
vector_length = torch.numel(x)
vector_sum = x.sum()

assert vector_length == 4
assert vector_sum.item() == 10.0
````

````python
# 문제 3. 행렬 A의 전치를 구하세요.
A = torch.arange(6).reshape(2, 3)
A_transposed = A.T

assert A_transposed.shape == (3, 2)
assert torch.equal(A_transposed, torch.tensor([[0, 3], [1, 4], [2, 5]]))
````

````python
# 문제 4. 3차 텐서를 만들고 두 번째 행렬을 선택하세요.
X = torch.arange(24, dtype=torch.float32).reshape(2,3,4)
second_matrix = X[1]

assert X.shape == (2, 3, 4)
assert second_matrix.shape == (3, 4)
assert second_matrix[0, 0].item() == 12
````

````python
# 문제 5. 같은 shape의 두 행렬에 Hadamard product를 적용하세요.
A = torch.tensor([[1.0, 2.0], [3.0, 4.0]])
B = torch.tensor([[2.0, 3.0], [4.0, 5.0]])
hadamard_product = A * B

assert torch.equal(hadamard_product, torch.tensor([[2.0, 6.0], [12.0, 20.0]]))
````

````python
# 문제 6. 전체 합, 열별 합, 행별 평균을 구하세요.
A = torch.arange(6, dtype=torch.float32).reshape(2, 3)
total = A.sum()
column_sums = A.sum(dim=0)
row_means = A.sum(dim=1) / 3

assert total.item() == 15.0
assert column_sums.tolist() == [3.0, 5.0, 7.0]
assert row_means.tolist() == [1.0, 4.0]
````

````python
# 문제 7. 행별 합을 구하되 축을 유지하고, 원래 행렬을 정규화하세요.
A = torch.tensor([[1.0, 2.0, 3.0], [2.0, 3.0, 5.0]])
row_sums = A.sum(
    dim=1,
    keepdim=True,
)
normalized = A / row_sums

assert row_sums.shape == (2, 1)
assert torch.allclose(normalized.sum(dim=1), torch.ones(2))
````

````python
# 문제 8. 두 벡터의 내적을 구하세요.
x = torch.tensor([1.0, 2.0, 3.0])
y = torch.tensor([4.0, 5.0, 6.0])
dot_product = torch.dot(x, y)

assert dot_product.shape == ()
assert dot_product.item() == 32.0
````

````python
# 문제 9. 행렬-벡터 곱을 계산하세요.
A = torch.tensor([[1.0, 2.0], [3.0, 4.0]])
x = torch.tensor([5.0, 6.0])
y = A @ x

assert y.shape == (2,)
assert torch.equal(y, torch.tensor([17.0, 39.0]))
````

````python
# 문제 10. 행렬-행렬 곱을 계산하세요.
A = torch.arange(6, dtype=torch.float32).reshape(2, 3)
B = torch.arange(12, dtype=torch.float32).reshape(3, 4)
C = A @ B

assert C.shape == (2, 4)
assert torch.equal(C, torch.tensor([[20.0, 23.0, 26.0, 29.0], [56.0, 68.0, 80.0, 92.0]]))
````

````python
# 문제 11. 벡터의 L1/L2 norm과 행렬의 Frobenius norm을 구하세요.
x = torch.tensor([3.0, -4.0])
A = torch.tensor([[1.0, 2.0], [2.0, 1.0]])

l1_norm = torch.linalg.vector_norm(x, ord=1)
l2_norm = torch.linalg.vector_norm(x, ord=2)
frobenius_norm = torch.linalg.matrix_norm(A, ord="fro")

assert l1_norm.item() == 7.0
assert l2_norm.item() == 5.0
assert torch.allclose(frobenius_norm, torch.sqrt(torch.tensor(10.0)))
````
