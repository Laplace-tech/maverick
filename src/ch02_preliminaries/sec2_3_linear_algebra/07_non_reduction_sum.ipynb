```python
import torch

A = torch.arange(
    6,
    dtype=torch.float32,
).reshape(2, 3)

print(A)
print("Shape:", A.shape)
```

```python
# keepdim=False 이면 축약한 축이 사라진다.
row_sums = A.sum(dim=1)

row_sums_keepdim = A.sum(
    dim=1,
    keepdim=True,
)

print("Without keepdim:")
print(row_sums)
print("Shape:", row_sums.shape)

print("\nWith keepdim:")
print(row_sums_keepdim)
print("Shape:", row_sums_keepdim.shape)

assert row_sums.shape == (2,)
assert row_sums_keepdim.shape == (2, 1)
```

```python
# [2, 3] / [2, 1] broadcasting으로 각 행을 자신의 합으로 나눈다.
normalized_A = A / row_sums_keepdim

print(normalized_A)

print("\nRow sums:")
print(normalized_A.sum(dim=1))
```

```python
# cumsum은 지정한 축을 따라 값을 누적하지만 shape를 축약하지 않는다.
cumulative_sum = A.cumsum(dim=0)

print("Original A:")
print(A)

print("\nCumulative sum along dim=0:")
print(cumulative_sum)
print("Shape:", cumulative_sum.shape)
```

```python
scores = torch.tensor([
    [1.0, 2.0, 3.0],
    [2.0, 2.0, 4.0],
])

denominator = scores.sum(
    dim=-1,
    keepdim=True,
)

weights = scores / denominator

print("Denominator shape:", denominator.shape)
print("Weights:")
print(weights)

print("\nWeight sums:")
print(weights.sum(dim=-1))
```
