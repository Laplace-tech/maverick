```python
import torch

x = torch.arange(3, dtype=torch.float32)
y = torch.ones(3, dtype=torch.float32)

print("x:", x)
print("y:", y)
```

```python
# 내적은 같은 위치의 원소를 곱한 다음 모두 더한다.
dot_product = torch.dot(x, y)
manual_dot_product = (x * y).sum()

print("torch.dot:", dot_product)
print("Elementwise product:", x * y)
print("Manual dot product:", manual_dot_product)
print("Shape:", dot_product.shape)
```

```python
# 가중치의 합이 1이면 내적은 가중평균을 나타낸다.
values = torch.tensor([10.0, 20.0, 30.0])
weights = torch.tensor([0.2, 0.3, 0.5])

weighted_average = torch.dot(values, weights)

print("Values:", values)
print("Weights:", weights)
print("Weight sum:", weights.sum())
print("Weighted average:", weighted_average)
```

```python
x = torch.tensor(
    [1.0, 2.0, 3.0],
    requires_grad=True,
)

y = torch.tensor([4.0, 5.0, 6.0])

score = torch.dot(x, y)
score.backward()

print("Score:", score)
print("Gradient of x:", x.grad)

# score = Σxᵢyᵢ이므로 ∂score/∂xᵢ = yᵢ다.
assert torch.equal(x.grad, y)
```

```python
# Attention에서는 query와 key의 내적으로 두 벡터의 관련도를 계산한다.
query = torch.tensor([1.0, 0.0, 1.0, 0.0])
key = torch.tensor([1.0, 1.0, 0.0, 0.0])

attention_score = torch.dot(query, key)

print("Query shape:", query.shape)
print("Key shape:", key.shape)
print("Attention score:", attention_score)
print("Score shape:", attention_score.shape)
```
