```python
import torch
```

```python
# 소수점이 포함된 값으로 생성하면 기본적으로 실수형 스칼라 텐서가 된다.
# 핵심: 스칼라는 shape가 ()인 0차원 텐서이며 사칙연산 결과도 스칼라 텐서다.
x = torch.tensor(3.0)
y = torch.tensor(2.0)

addition = x + y
multiplication = x * y
division = x / y
exponentiation = x ** y

print("x + y:", addition)
print("x * y:", multiplication)
print("x / y:", division)
print("x ** y:", exponentiation)

assert addition.item() == 5.0
assert multiplication.item() == 6.0
assert division.item() == 1.5
assert exponentiation.item() == 9.0
```

```python
# 스칼라 텐서의 shape는 빈 튜플이며 축이 존재하지 않는다.
# 핵심: shape ()인 스칼라와 shape (1,)인 원소 하나짜리 벡터는 축 개수가 서로 다르다.
scalar = torch.tensor(3.0)
one_element_vector = torch.tensor([3.0])

print("Scalar:")
print("Value:", scalar) # 3.0
print("Shape:", scalar.shape) # []
print("Number of axes:", scalar.ndim) # 0
print("Number of elements:", scalar.numel()) # 1

print("\nOne-element vector:")
print("Value:", one_element_vector) # 3.0
print("Shape:", one_element_vector.shape) # [1]
print("Number of axes:", one_element_vector.ndim) # 1

assert scalar.shape == ()
assert one_element_vector.shape == (1,)
```

```python
# 딥러닝의 loss 역시 일반적으로 0차원 스칼라 텐서
# 핵심: 딥러닝의 loss는 보통 스칼라이며 backward는 loss에서 출발해 계산 그래프를 역방향으로 미분한다.
weight = torch.tensor(2.0, requires_grad=True)
loss = (weight - 5) ** 2

# dloss/dweight = 2 * (weight-5)이므로,  weight=2에서 gradient는 -6.
loss.backward()

print("Loss:", loss)
print("Loss shape", loss.shape)
print("Gradient:", weight.grad)
```
