```python
import torch

u = torch.tensor([3.0, -4.0])

print("Vector:", u)
print("Shape:", u.shape)
```

```python
# L2 norm은 제곱합의 제곱근이다.
l2_norm = torch.linalg.vector_norm(u, ord=2)
manual_l2_norm = torch.sqrt((u ** 2).sum())

print("L2 norm:", l2_norm)
print("Manual L2 norm:", manual_l2_norm)
print("Shape:", l2_norm.shape)

assert l2_norm.item() == 5.0
assert l2_norm == manual_l2_norm
assert l2_norm.shape == ()
```

```python
# L1 norm은 절댓값의 합이다.

l1_norm = torch.linalg.vector_norm(u, ord=1)
manual_l1_norm = torch.abs(u).sum()

print("L1 norm:", l1_norm)
print("Manual L1 norm:", manual_l1_norm)

assert l1_norm.item() == 7.0
assert l1_norm == manual_l1_norm
```

```python
# 일반적인 Lp norm을 직접 계산한다.
p = 3

lp_norm = (
    torch.abs(u).pow(p).sum()
).pow(1 / p)

library_lp_norm = torch.linalg.vector_norm(
    u,
    ord=p,
)

print("Manual L3 norm:", lp_norm)
print("Library L3 norm:", library_lp_norm)
```

```python
# Frobenius norm은 행렬 전체 원소에 대한 L2 norm과 같다.
X = torch.ones((4, 9))

frobenius_norm = torch.linalg.matrix_norm(
    X,
    ord="fro",
)

manual_frobenius_norm = torch.sqrt(
    (X.pow(2).sum())
)

print("Frobenius norm:", frobenius_norm)
print("Manual norm:", manual_frobenius_norm)
```

```python
# L2 norm의 gradient는 x / ‖x‖₂다.
x = torch.tensor(
    [3.0, 4.0],
    requires_grad=True,
)

length = torch.linalg.vector_norm(x)
length.backward()

print("Length:", length)
print("Gradient:", x.grad)

expected_gradient = torch.tensor([
    3.0 / 5.0,
    4.0 / 5.0,
])

assert torch.allclose(
    x.grad,
    expected_gradient,
)
```

```python
# 제곱 L2 norm은 제곱근이 없어 gradient가 단순히 2x가 된다.

x = torch.tensor(
    [3.0, 4.0],
    requires_grad=True,
)

squared_l2_norm = x.pow(2).sum()
squared_l2_norm.backward()

print("Squared L2 norm:", squared_l2_norm)
print("Gradient:", x.grad)
```
