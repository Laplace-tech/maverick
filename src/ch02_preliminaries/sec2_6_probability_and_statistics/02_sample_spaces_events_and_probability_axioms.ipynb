```python
import torch
```

```python
sample_space = torch.arange(1, 7)

outcome_probabilities = torch.full(
    size=(6,),
    fill_value = 1 / 6,
)

print("sample space:", sample_space)
print("outcome probabilities:", outcome_probabilities)
print("total probability:", outcome_probabilities.sum())

assert sample_space.shape == (6,)
assert outcome_probabilities.shape == (6,)
assert torch.isclose(
    outcome_probabilities.sum(),
    torch.tensor(1.0),
)
```

```python
def probability(event_mask):
    # Boolean mask로 표현된 사건의 확률을 합산한다
    return outcome_probabilities[event_mask].sum()

# 사건 A: 주사위 결과가 5다.
event_A = sample_space == 5

# 사건 B: 주사위 결과가 홀수다.
event_B = sample_space % 2 == 1

probability_A = probability(event_A)
probability_B = probability(event_B)

print("Event A outcomes:", sample_space[event_A])
print("P(A):", probability_A)

print("\nEvent B outcomes:", sample_space[event_B])
print("P(B):", probability_B)
```

```python
# 집합의 교집합 A∩B는 Boolean AND로 표현한다.
intersection = event_A & event_B

# 집합의 합집합 A∪B는 Boolean OR로 표현한다.
union = event_A | event_B

print("A ∩ B:", sample_space[intersection])
print("A ∪ B:", sample_space[union])
```

```python
# 사건 B의 여사건은 “홀수가 아니다”, 즉 짝수가 나오는 사건이다.
event_B_complement = ~event_B

# B와 Bᶜ의 합집합은 전체 표본공간이다.
entire_space = event_B | event_B_complement

# B와 Bᶜ의 교집합은 공집합이다.
empty_event = event_B & event_B_complement

print("B outcomes:", sample_space[event_B])
print("B complement:", sample_space[event_B_complement])
print("P(B):", probability(event_B))
print("P(B complement):", probability(event_B_complement))
```
