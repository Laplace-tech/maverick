````python
import torch
````

````python
# 문제 1. f(x)=3x²-4x의 도함수를 x=2에서 계산하세요.
x = torch.tensor(2.0)
derivative = 6 * x - 4;

assert derivative.shape == ()
assert derivative.item() == 8.0
````

````python
# 문제 2. 중앙 차분으로 f'(1)을 근사하세요.
def f(x):
    return 3 * x**2 - 4 * x


x = torch.tensor(1.0, dtype=torch.float64)
h = torch.tensor(1e-4, dtype=torch.float64)
numerical_derivative = (f(x+h) - f(x-h)) / (2 * h)

assert torch.isclose(
    numerical_derivative,
    torch.tensor(2.0, dtype=torch.float64),
    atol=1e-8,
)
````

````python
# 문제 3. x₀=1에서의 접선 T(x)=f(x₀)+f'(x₀)(x-x₀)를 계산하세요.
x_values = torch.tensor([0.0, 1.0, 2.0])
x0 = torch.tensor(1.0)
function_value_at_x0 = f(x0)
slope_at_x0 = 6 * x0 - 4
tangent_values = (slope_at_x0) * (x_values - x0) + f(x0)

assert torch.equal(tangent_values, torch.tensor([-3.0, -1.0, 1.0]))
````

````python
# 문제 4. f(x₁,x₂)=3x₁²+5eˣ²의 gradient를 x=(2,0)에서 구하세요.
x = torch.tensor([2.0, 0.0])
partial_x1 = 6 * x[0]
partial_x2 = 5 * torch.exp(x[1])
gradient = torch.stack((partial_x1, partial_x2))

assert gradient.shape == x.shape
assert torch.equal(gradient, torch.tensor([12.0, 5.0]))
````

````python
# 문제 5. q(x)=xᵀAx의 gradient (A+Aᵀ)x를 계산하세요.
A = torch.tensor([[1.0, 2.0], [3.0, 4.0]])
x = torch.tensor([1.0, 2.0])
quadratic_gradient = (A + A.T) @ x

assert quadratic_gradient.shape == x.shape
assert torch.equal(quadratic_gradient, torch.tensor([12.0, 21.0]))
````

````python
# 문제 6. u=x², y=sin(u)에 연쇄 법칙을 적용하세요.
x = torch.tensor(2.0)
u = x**2
dy_du = torch.cos(u)
du_dx = 2 * x
dy_dx = dy_du * du_dx

assert torch.isclose(dy_dx, 2 * x * torch.cos(x**2))
````

````python
# 문제 7. ∇ₓy=A∇ᵤy를 계산하세요.
A = torch.tensor([[1.0, 2.0], [1.0, 1.0]])
gradient_u = torch.tensor([6.0, 3.0])
gradient_x = A @ gradient_u

assert gradient_x.shape == (2,)
assert torch.equal(gradient_x, torch.tensor([12.0, 9.0]))
````
