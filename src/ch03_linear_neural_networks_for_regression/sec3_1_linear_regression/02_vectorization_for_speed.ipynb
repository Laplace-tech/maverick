```python
import time
import torch

torch.manual_seed(42)

n = 10_000
a = torch.ones(n)
b = torch.ones(n)

print("a.shape:", a.shape)
print("b.shape:", b.shape)
```

```python
loop_result = torch.zeros(n)

start = time.perf_counter()

for i in range(n):
    loop_result[i] = a[i] + b[i]
    
loop_elapsed = time.perf_counter() - start

print(f"Python for-loop: {loop_elapsed:.6f} sec")
print("loop_result.shape:", loop_result.shape)
```

```python
start = time.perf_counter()

vectorized_result = a + b

vectorized_elapsed = time.perf_counter() - start

print(f"Vectorized operation: {vectorized_elapsed:.6f} sec")
print(f"Speedup: {loop_elapsed / vectorized_elapsed:.1f}x")
print("vectorized_result.shape:", vectorized_result.shape)

# 두 방식의 수학적 결과는 반드시 같아야 한다.
assert torch.equal(loop_result, vectorized_result)
```

```python
# 선형 회귀에서의 벡터화:
# features: (batch_size, num_features) = (4, 2)
# weights:  (num_features, 1)        = (2, 1)
# output:   (batch_size, 1)          = (4, 1)

features = torch.tensor([
    [1.0, 2.0],
    [2.0, 1.0],
    [3.0, 4.0],
    [4.0, 3.0],
])

weights = torch.tensor([
    [2.0],
    [-3.0],
])

bias = torch.tensor([5.0])

# 샘플 하나씩 predicting 하는 방식
loop_predictions = torch.stack([
    sample @ weights + bias
    for sample in features
])

# 미니배치 전체를 행렬곱으로 prediction 하는 방식
vectorized_predictions = features @ weights + bias

print("Loop predictions:\n", loop_predictions)
print("\nVectorized predictions:\n", vectorized_predictions)

assert loop_predictions.shape == (4, 1)
assert torch.equal(loop_predictions, vectorized_predictions)
```
