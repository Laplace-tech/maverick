```python
import torch
```

```python
# 실제 데이터 분포:
# y = 2x + 1 + epsilon
# epsilon ~ N(0, 0.5^2)

true_weight = 2.0
true_bias = 1.0
noise_std = 0.5

def sample_data(number_of_examples):
    """동일한 분포에서 IID sample을 생성"""

    # [N, 1] 차원    
    features = torch.randn(
        number_of_examples,
        1,
    )
    
    # [N, 1] 차원    
    noise = (
        torch.randn(number_of_examples, 1) 
        * noise_std # 노이즈 표준편차
    )
    
    labels = (
        features * true_weight
        + true_bias
        + noise
    )
    
    return features, labels


def fixed_model(features):
    return features * true_weight + true_bias

def mean_squared_error(predictions, labels):
    return (predictions - labels).pow(2).mean()
```

```python
# 유한한 training set에서 empirical error를 계산
train_features, train_label = sample_data(
    number_of_examples=100_000,
)

train_predictions = fixed_model(train_features)

training_error = mean_squared_error(
    train_predictions,
    train_label,
)

# 매우 큰 독립 test set을 이용해
# population generalization error를 근사
test_features, test_labels = sample_data(
    number_of_examples=100_000,
)

test_predictions = fixed_model(test_features)

estimated_generalization_error = mean_squared_error(
    test_predictions,
    test_labels,
)

generalization_gap = (
    estimated_generalization_error
    - training_error
)

print("Training error:", training_error.item())
print(
    "Estimated generalization error:",
    estimated_generalization_error.item(),
)
print("Generalization gap:", generalization_gap.item())
```

```python
# Cell 4
# Sample이 많아질수록 empirical error는
# population error인 0.25 주변에서 더 안정적으로 측정된다.
sample_sizes = [10, 100, 1_000, 10_000, 100_000, 1_000_000]

for sample_size in sample_sizes:
    features, labels = sample_data(sample_size)
    predictions = fixed_model(features)

    empirical_error = mean_squared_error(
        predictions,
        labels,
    )

    print(
        f"n={sample_size:5d} "
        f"| empirical error={empirical_error.item():.4f}"
    )
```
