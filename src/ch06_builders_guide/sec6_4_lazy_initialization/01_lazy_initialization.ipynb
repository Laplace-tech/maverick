```python
import torch
from torch import nn
from d2l import torch as d2l

# Lazy Initialization을 사용하면 parameter shape을 자동으로 추론함
```

```python
# Initialization 이전의 Lazy Parameter 확인

net = nn.Sequential(
    nn.LazyLinear(256),
    nn.ReLU(),
    nn.LazyLinear(10),
)

first_layer = net[0]

assert isinstance(
    first_layer,
    nn.LazyLinear,
)

# <Error>: Can't access the shape of an uninitialized parameter
# print("first_layer.weight.shape", first_layer.weight.shape)
print(first_layer.weight)
```

```python
# 최초 Forward Propagation으로 parameter initialization

# [2, 20]
X = torch.rand(2, 20)

# [2, 20] -> [2, 256] -> [2, 10]
Y = net(X)

print("Y.shape", Y.shape)
print("\nfirst_layer.weight.shape:", first_layer.weight.shape)
print(first_layer.weight)
```

```python
# Dummy Input을 사용하는 apply_init Method


def module_apply_init(
    self,
    inputs,
    init=None,
):

    # Dry run으로 Lazy Parameter shape 결정
    self.forward(*inputs)

    # Shape 결정 후 custom initializer 적용
    if init is not None:
        self.net.apply(init)


d2l.Module.apply_init = module_apply_init
```
