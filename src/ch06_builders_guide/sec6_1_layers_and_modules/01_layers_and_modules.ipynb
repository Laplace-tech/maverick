```python
# nn.Sequential MLP 구성

import torch
from torch import nn
from torch.nn import functional as F
```

```python
net = nn.Sequential(
    nn.LazyLinear(256), # [2, 20] -> [2, 256]
    nn.ReLU(),          # [2, 256] -> [2, 256]
    nn.LazyLinear(10),  # [2, 256] -> [2, 10]
)

# [2, 20]
X = torch.rand(2, 20)

print(net)
print(net(X).shape) # [2, 10]
```

```python
# Custom Module로 MLP 정의

class MLP(nn.Module):

    def __init__(self):

        super().__init__()

        self.hidden = nn.LazyLinear(256)
        self.out = nn.LazyLinear(10)


def mlp_forward(self, X):

    # [B, 20] -> [B, 256]
    H = F.relu(
        self.hidden(X)
    )

    # [B, 256] -> [B, 10]
    return self.out(H)


MLP.forward = mlp_forward
```

```python
# Custom Module 실행 확인

net = MLP()

Y = net(X)

print(net)
print(Y.shape)
```
