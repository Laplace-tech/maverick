```python
import torch
from torch import nn
from torch.nn import functional as F
```

```python
# MySequential 직접 구현

class MySequential(nn.Module):
    
    def __init__(
        self,
        *args,
    ):
        
        super().__init__()
        
        for idx, module in enumerate(args):
            
            self.add_module(
                str(idx),
                module,
            )
            
def my_sequential_forward(self, X):
    
    for module in self.children():
    
        X = module(X)

    return X

MySequential.forward = my_sequential_forward
```

```python
# MySequential로 MLP 구성

# [2, 20]
X = torch.rand(2, 20)

net = MySequential(
    nn.LazyLinear(256), # [B, 20] -> [B, 256]
    nn.ReLU(),         # [B, 256] -> [B, 256]
    nn.LazyLinear(10)  # [B, 256] -> [B, 10]
)

Y = net(X)

print(net)
```
