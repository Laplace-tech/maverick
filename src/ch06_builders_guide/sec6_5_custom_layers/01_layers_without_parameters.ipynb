```python
# 셀 1 — 기본 import

import torch
from torch import nn
from torch.nn import functional as F
from d2l import torch as d2l
```

```python
class CenteredLayer(nn.Module):
    
    def __init__(self):
        super().__init__()
        
def centered_layer_forward(
    self,
    X,
):
    
    return X - X.mean()

CenteredLayer.forward = centered_layer_forward
```

```python
# CenteredLayer 단독 실행

layer = CenteredLayer()

X = torch.tensor(
    [1.0, 2, 3, 4, 5]
)

Y = layer(X)

print(Y)

# 빈 list는 이 layer에 trainable parameter가 없다는 뜻.
print(
    list(
        layer.parameters()
    )
)
```

```python
# Sequential 내부에서 Custom Layer 사용

net = nn.Sequential(
    nn.LazyLinear(128),
    CenteredLayer(),
)

# [4, 8] -> [4, 128]
X = torch.rand(4, 8)
Y = net(X)

print(Y.shape)
print(Y.mean())
```
