```python
import torch
from torch import nn

net = nn.Sequential(
    nn.LazyLinear(8), # [2, 4] -> [2, 8]: Weight, bias
    nn.ReLU(),        # [2, 8] -> [2, 8]: activation function
    nn.LazyLinear(1), # [2, 8] -> [2, 1]: Weight, bias
)

# [2, 4]
X = torch.rand(
    size=(2, 4)
)

# Lazy parameter initialization
Y = net(X)

print(Y.shape)
```

```python
# 특정 layer의 state_dict 확인

output_layer_state = net[0].state_dict()

print(output_layer_state)
```

```python
# 특정 parameter의 value와 gradient 확인

bias = net[2].bias

print(type(bias))
print(bias.data)
print(net[2].weight.grad is None)
```

```python
# 모든 Parameter의 이름과 shape 확인

parameter_shapes = [
    (
        name,
        parameter.shape,
    )
    for name, parameter
    in net.named_parameters()
]

print(parameter_shapes)
```
