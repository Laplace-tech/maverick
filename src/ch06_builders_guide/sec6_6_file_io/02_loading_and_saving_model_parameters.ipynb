```python
from pathlib import Path

import torch
from torch import nn
from torch.nn import functional as F
from d2l import torch as d2l
```

```python
# 저장할 MLP 정의

class MLP(nn.Module):
    
    def __init__(self):
        super().__init__()
        
        self.hidden = nn.LazyLinear(256) # [B, input] -> [B, 256]
        self.output = nn.LazyLinear(10)  # [B, 256] -> [B, 10]
        
def mlp_forward(
    self,
    X,
):
    
    # [B, 20] -> [B, 256]
    H = F.relu(
        self.hidden(X)
    )
    
    # [B, 256] -> [B, 10]
    return self.output(H)

MLP.forward = mlp_forward
```

```python
# Original Model의 Parameter 저장

net = MLP()

# [2, 20]
X = torch.randn(
    [2, 20]
)

# Lazy parameter materialization
Y = net(X)

data_dir = Path("data")
data_dir.mkdir(
    exist_ok=True
)

model_path = (
    data_dir
    / "mlp.params"
)

# 현재 모델의 파라미터 값을 Dictionary 형태로 save
torch.save(
    net.state_dict(),
    model_path,
)

print(net)
print(Y.shape)
print(model_path)
```

```python
# 새 Model에 저장된 Parameter 불러오기

clone = MLP()

state_dict = torch.load(
    model_path,
    weights_only=True,
)

load_result = clone.load_state_dict(
    state_dict
)

print(clone)
print(load_result)
```

```python
# Original Model vs Clone

Y_clone = clone(X)

print(
    Y_clone == Y
)

print(
    torch.equal(
        Y_clone,
        Y,
    )
)

print(
    torch.allclose(
        Y_clone,
        Y,
    )
)
```
