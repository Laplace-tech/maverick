```python
# GPU에 Tensor 직접 생성

assert torch.cuda.is_available()

gpu_device = torch.device(
    "cuda:0"
)

# GPU로 지정
X = torch.ones(
    (2, 3),
    device=gpu_device,
)

# CPU로 지정
Y = torch.rand(
    (2, 3),
)

print(X)
print(Y)
print(X.device)
print(Y.device)
```

```python
# 서로 다른 Device의 연산 실패 확인:
# all tensors have to be on the same device!!!

try:
    X + Y

except RuntimeError as error:
    print(type(error).__name__)
    print(error)
```

```python
# 같은 Device로 이동한 뒤 연산

Z = Y.to(
    X.device
)

result = X + Z

print(Z)
print(Z.device)
print(result)
```

```python
Z_again = Z.to(
    X.device
)

print(
    Z_again is Z
)

X_numpy = (
    X
    .cpu()
    .numpy()
)

print(X_numpy)
```
