```python
import torch
from torch import nn
from d2l import torch as d2l
```

```python
# CPU & GPU Device 생성

def cpu():
    
    return torch.device(
        "cpu"
    )
    
def gpu(
    index=0,
):

    return torch.device(
        f"cuda:{index}"
    )
    
print(cpu())
print(gpu())
print(gpu(1))
```

```python
# 사용 가능한 GPU 확인

def num_gpus():
    
    return (
        torch.cuda.device_count()
    )
    
print(
    torch.cuda.is_available()
)

print(
    num_gpus()
)

if torch.cuda.is_available():
    
    print(
        torch.cuda.get_device_name(0)
    )
    
    properties = (
        torch.cuda.get_device_properties(0)
    )
    
    total_memory_gb = (
        properties.total_memory
        / 1024**3
    )
    
    print(
        f"{total_memory_gb:.1f} GB"
    )
```

```python
# 존재하는 GPU만 안전하게 선택

def try_gpu(
    index=0,
):

    if num_gpus() >= index + 1:

        return gpu(index)

    return cpu()


def try_all_gpus():

    if num_gpus() == 0:

        return [cpu()]

    return [
        gpu(index)
        for index
        in range(num_gpus())
    ]


print(
    try_gpu()
)
print(
    try_gpu(10)
)
print(
    try_all_gpus()
)
```
