```python
import torch
from torch import nn
from d2l import torch as d2l
```

```python
# Concise Bidirectional GRU
#
# Input       : [T, B, D]
# Output      : [T, B, 2H]
# Final State : [2L, B, H]

class BiGRU(d2l.RNN):
    def __init__(
        self,
        num_inputs: int,
        num_hiddens: int,
    ) -> None:
        d2l.Module.__init__(self)

        self.num_inputs = num_inputs
        self.single_direction_hiddens = num_hiddens
        
        # 두 Direction의 output이 concatenate 되므로 
        # model의 최종 output dimension은 2H
        self.num_hiddens = 2 * num_hiddens
        
        self.rnn = nn.GRU(
            input_size=num_inputs,
            hidden_size=num_hiddens,
            bidirectional=True,
        )
```

```python
# Tensor Shape와 Parameter 검증

num_steps = 3
batch_size = 2
num_inputs = 4
num_hiddens = 8

# X: [T, B, D]
X = torch.randn(
    num_steps,
    batch_size,
    num_inputs,
)

bigru = BiGRU(
    num_inputs=num_inputs,
    num_hiddens=num_hiddens,
)

#       Y    : [T, B, 2H]
# final_state: [2 * num_layers, B, H]
Y, final_state = bigru.rnn(X)

num_parameters = sum(
    parameter.numel()
    for parameter in bigru.parameters()
)

print("Input shape:", tuple(X.shape))
print("Output shape:", tuple(Y.shape))
print(
    "Final state shape:",
    tuple(final_state.shape),
)
print("Parameter count:", num_parameters)
```
