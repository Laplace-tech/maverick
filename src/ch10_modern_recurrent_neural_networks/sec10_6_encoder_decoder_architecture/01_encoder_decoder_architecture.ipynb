```python
import torch
from torch import nn
from d2l import torch as d2l

# Encoder-Decoder Architecture:
# input과 output이 모두 가변 길이인 Sequence-to-Sequence 문제를 처리
```

```python
# Encoder Interface:
# 가변 길이 Input sequence를 받아 encoded state로 변환

class Encoder(nn.Module):

    def __init__(self) -> None:
        super().__init__()
        
    def forward(
        self,
        X: torch.Tensor,
        *args: object,
    ) -> object: 
        raise NotImplementedError
```

```python
# Decoder Interface

class Decoder(nn.Module):
    
    def __init__(self) -> None:
        super().__init__()
    
    # Encoder output을 Decoder가 사용할 encoded state로 변환
    def init_state(
        self,
        enc_all_outputs: object,
        *args: object,
    ) -> object:
        raise NotImplementedError
    
    # target input과 encoded state로 다음 output 계산
    def forward(
        self,
        X: torch.Tensor,
        state: object,
    ) -> tuple[
        torch.Tensor,
        object,
    ]:
        raise NotImplementedError
```

```python
# Encoder-Decoder Architecture

class EncoderDecoder(d2l.Classifier):
    
    def __init__(
        self,
        encoder: Encoder,
        decoder: Decoder,
    ) -> None:
        super().__init__()
        
        self.encoder = encoder
        self.decoder = decoder

    def forward(
        self,
        enc_X: torch.Tensor, # source sequence
        dec_X: torch.Tensor, # <bos>로 시작하는 target input
        *args: object,
    ) -> torch.Tensor:
        
        # source sequence를 encode
        enc_all_outputs = self.encoder(
            enc_X,
            *args,
        )
        
        # decoder가 encoder output을 decoder state로 변환
        dec_state = self.decoder.init_state(
            enc_all_outputs,
            *args,
        )
        
        # target input과 state로 다음 output을 계산
        decoder_outputs, _ = self.decoder(
            dec_X,
            dec_state,
        )
        
        return decoder_outputs
```
