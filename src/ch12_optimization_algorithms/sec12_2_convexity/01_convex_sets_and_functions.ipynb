```python
%matplotlib inline

import numpy as np
import torch
from torch import nn
from d2l import torch as d2l
```

```python
# Convex Function & Nonconvex Function

def quadratic(
    x: torch.Tensor,
) -> torch.Tensor:
    return 0.5 * x.pow(2)

def cosine(
    x: torch.Tensor,
) -> torch.Tensor:
    return torch.cos(
        np.pi * x
    )
    
def exponential(
    x: torch.Tensor,
) -> torch.Tensor:
    return torch.exp(
        0.5 * x
    )
    
x = torch.arange(
    -2.0,
    2.0,
    0.01,
)

segment = torch.tensor(
    [-1.5, 1.0]
)

functions = (
    quadratic,
    cosine,
    exponential,
)

titles = (
    "Convex: Quadratic",
    "Nonconvex: Cosine",
    "Convex: Exponential",
)


_, axes = d2l.plt.subplots(
    1,
    3,
    figsize=(9, 3),
)

for axis, function, title in zip(
    axes,
    functions,
    titles,
    strict=True,
):
    d2l.plot(
        [x, segment],
        [
            function(x),
            function(segment),
        ],
        axes=axis,
    )

    axis.set_title(title)
```
