```python
%matplotlib inline

import torch
from torch import nn
from d2l import torch as d2l
```

```python
# Local Minimum & Global Minimum

def convex_function(
    x: torch.Tensor,
) -> torch.Tensor:
    return (
        x - 1
    ) ** 2
    
x = torch.arange(
    -2.0,
    2.0,
    0.01,
)

segment = torch.tensor(
    [-1.5, 1.0]
)

d2l.plot(
    [
        x,
        segment,
    ],
    [
        convex_function(x),
        convex_function(segment),
    ],
    xlabel="x",
    ylabel="f(x)",
    legend=[
        "f(x)",
        "line segment",
    ],
)
```

```python
# Hessian & Eigenvalue로 Convexity 확인

def quadratic_2d(
    x: torch.Tensor,
) -> torch.Tensor:
    return 0.5 * (
        x @ x
    )
    
point = torch.tensor(
    [2.0, -3.0],
    requires_grad=True,
)

hessian_matrix = torch.autograd.functional.hessian(
    quadratic_2d,
    point,
)

eigenvalues = torch.linalg.eigvalsh(
    hessian_matrix
)

print(
    "Hessian:"
)

print(
    hessian_matrix
)

print(
    "\nEigenvalues:",
    eigenvalues,
)

print(
    "\nPositive semidefinite:",
    bool(
        torch.all(
            eigenvalues >= 0
        )
    ),
)
```
