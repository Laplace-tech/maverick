```python
%matplotlib inline

import torch
from torch import nn
from d2l import torch as d2l
```

```python
# Past Gradient Weight

momentum_coefficients = [
    0.95,
    0.90,
    0.60,
    0.00,
]

time_offsets = torch.arange(
    40,
    dtype=torch.float32,
)

figure, axis = d2l.plt.subplots(
    figsize=(
        7,
        4,
    )
)

for coefficient in momentum_coefficients:
    gradient_weights = (
        coefficient ** time_offsets
    )

    axis.plot(
        time_offsets,
        gradient_weights,
        linewidth=2,
        label=(
            f"beta = {coefficient:.2f}"
        ),
    )

axis.set_xlabel(
    "Steps into the past"
)

axis.set_ylabel(
    "Gradient weight"
)

axis.set_title(
    "Momentum Weight Decay"
)

axis.legend()

axis.grid(
    alpha=0.3,
)

figure.tight_layout()

d2l.plt.show()
```

```python
# Effective Total Weight

for coefficient in momentum_coefficients:
    effective_total_weight = (
        1
        / (1 - coefficient)
    )

    print(
        f"beta={coefficient:.2f} | "
        f"total weight="
        f"{effective_total_weight:.2f}"
    )
```
