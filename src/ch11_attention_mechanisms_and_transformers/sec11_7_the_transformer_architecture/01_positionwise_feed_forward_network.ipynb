```python
import torch
from torch import nn
from d2l import torch as d2l
```

```python
# Position-Wise Feed-Forward Network

class PositionWiseFFN(nn.Module):
    
    def __init__(
        self,
        ffn_num_hiddens: int,
        ffn_num_outputs: int,
    ) -> None:
        super().__init__()
        
        self.dense1 = nn.LazyLinear(
            out_features=ffn_num_hiddens,
        )
        
        self.relu = nn.ReLU()
        
        self.dense2 = nn.LazyLinear(
            out_features=ffn_num_outputs,
        )
    
        
    def forward(
        self,
        X: torch.Tensor,
    ) -> torch.Tensor:
        
        # [B, T, D] → [B, T, H_ffn] → [B, T, D_out]
        return self.dense2(
            self.relu(
                self.dense1(X)
            )
        )
```

```python
# Shape & Parameter sharing 검증

ffn = PositionWiseFFN(
    ffn_num_hiddens=4,
    ffn_num_outputs=8,
)

ffn.eval()

# [B, T, D] = [2, 3, 4]
X = torch.ones(
    (
        2,
        3,
        4,
    )
)

with torch.no_grad():
    Y = ffn(X)
    
    
print(
    "Input shape:",
    tuple(X.shape),
)

print(
    "Output shape:",
    tuple(Y.shape),
)

print(
    "\nY[0] example:",
)
print(
    Y[0]
)

print(
    "\nY[1] example:",
)
print(
    Y[1]
)
```
