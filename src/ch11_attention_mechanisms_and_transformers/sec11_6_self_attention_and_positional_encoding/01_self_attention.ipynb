```python
import torch
from torch import nn
from d2l import torch as d2l
```

```python
# Self-Attention Data Flow

num_hiddens = 100
num_heads = 5

batch_size = 2
num_steps = 4

valid_lens = torch.tensor([
    3,
    2,
])

attention = d2l.MultiHeadAttention(
    num_hiddens=num_hiddens,
    num_heads=num_heads,
    dropout=0.5,
)

# Dropout 비활성화
attention.eval()


# Token sequence:
# [B, T, D] = [2, 4, 100]
X = torch.ones(
    (
        batch_size,
        num_steps,
        num_hiddens,
    )
)

with torch.no_grad():
    
    # Self-Attention에서는 동일한 sequence X가
    # Query, Key, Value의 출발점으로 사용
    Y = attention(
        queries=X,
        keys=X,
        values=X,
        valid_lens=valid_lens
    )
    
print(
    "Input shape:",
    tuple(X.shape),
)

print(
    "Output shape:",
    tuple(Y.shape),
)

# Self-Attention은 sequence length를 유지한다.
d2l.check_shape(
    Y,
    (
        batch_size,
        num_steps,
        num_hiddens,
    ),
)
```
