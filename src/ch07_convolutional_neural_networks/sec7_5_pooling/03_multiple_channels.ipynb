```python
import torch
from torch import nn
from d2l import torch as d2l
```

```python
# Multiple Channel Pooling

X = torch.arange(
    16,
    dtype=torch.float32,
).reshape((1, 1, 4, 4))

# [1, 2, 4, 4]
X_multi = torch.cat(
    (
        X,
        X + 1,
    ),
    dim=1,
)

channel_pool = nn.MaxPool2d(
    kernel_size=3,
    padding=1,
    stride=2,
)

# Pooling은 Channel 수를 변경하지 않는다.
Y_multi = channel_pool(X_multi)

print("Input shape:", X_multi.shape)
print(X_multi)

print("\nOutput shape:", Y_multi.shape)
print(Y_multi)
```
