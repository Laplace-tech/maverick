```python
import torch
from torch import nn
from d2l import torch as d2l
```

```python
# Padding으로 input, output shape 유지

def comp_conv2d(
    conv2d: nn.Conv2d, 
    X: torch.Tensor, # [H, W]
) -> torch.Tensor:
    input_height, input_width = X.shape
    
    # X_4d: [1, 1, H, W]
    X_4d = X.reshape(
        (1, 1, input_height, input_width)
    )
    # Y_4d: [1, 1, H, W]
    Y_4d = conv2d(X_4d)
    
    # [H, W]
    return Y_4d[0, 0]
    
X = torch.rand(
    size=(8, 8)
)    

conv2d = nn.LazyConv2d(
    out_channels=1,
    kernel_size=3,
    padding=1, # p=k-1
)

Y = comp_conv2d(conv2d, X)

print("X shape:", X.shape)
print("Y shape:", Y.shape)
```

```python
# 서로 다른 Height와 Width Padding

conv2d = nn.LazyConv2d(
    out_channels=1,
    kernel_size=(5, 3),
    padding=(2, 1),
)

Y = comp_conv2d(conv2d, X)

print("X shape:", X.shape)
print("Y shape:", Y.shape)
```

```python
# Stride 2로 Resolution 절반 감소

conv2d = nn.LazyConv2d(
    out_channels=1,
    kernel_size=3,
    padding=1,
    stride=2,
)

Y = comp_conv2d(
    conv2d,
    X,
)

print("X shape:", X.shape)
print("Y shape:", Y.shape)
```

```python
# 서로 다른 Kernel, Padding, Stride

conv2d = nn.LazyConv2d(
    out_channels=1,
    kernel_size=(3, 5),
    padding=(0, 1),
    stride=(3, 4),
)

Y = comp_conv2d(
    conv2d,
    X,
)

print("X shape:", X.shape)
print("Y shape:", Y.shape)
```
