```python
import collections
import random
import re

import torch
from d2l import torch as d2l
```

```python
class TimeMachine(d2l.DataModule):
    """The Time Machine dataset."""

    def _download(self) -> str:
        filename = d2l.download(
            d2l.DATA_URL + "timemachine.txt",
            self.root,
            "090b5e7e70c295757f55df93cb0a180b9691891a",
        )

        with open(filename, encoding="utf-8") as file:
            return file.read()

    def _preprocess(self, text: str) -> str:
        return re.sub(
            r"[^A-Za-z]+",
            " ",
            text,
        ).lower()
```

```python
data = TimeMachine()
raw_text = data._download()

print(repr(raw_text[:60]))
print("Raw character count:", len(raw_text))
```

```python
text = data._preprocess(raw_text)

print(repr(text[:60]))
print("Processed character count:", len(text))
```

```python
def time_machine_tokenize(
    self: TimeMachine,
    text: str,
) -> list[str]:
    return list(text)


TimeMachine._tokenize = time_machine_tokenize
```

```python
tokens: list[str] = data._tokenize(text)

print(",".join(tokens[:30]))
print("First 30 tokens:", tokens[:30])
print("Token count:", len(tokens))
print("Text length:", len(text))

assert len(tokens) == len(text)
assert "".join(tokens) == text
```
