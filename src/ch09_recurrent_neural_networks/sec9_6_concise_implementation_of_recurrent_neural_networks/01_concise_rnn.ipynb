```python
import torch
from torch import nn
from torch.nn import functional as F
from d2l import torch as d2l
```

```python
# High-Level API RNN

# Input  : [T, B, V]
# Hidden : [T, B, H]
# Output : [B, T, V]
class RNN(d2l.Module):
    
    num_inputs: int
    num_hiddens: int
    
    def __init__(
        self,
        num_inputs: int,  # V
        num_hiddens: int, # H
    ) -> None:
        super().__init__()
        self.save_hyperparameters()
        
        self.rnn = nn.RNN(
            input_size=num_inputs,
            hidden_size=num_hiddens,
        )
        
    def forward(
        self,
        inputs: torch.Tensor,
        state: torch.Tensor | None = None,
    ) -> tuple[torch.Tensor, torch.Tensor]:
        
        # inputs:  [T, B, V]
        # outputs: [T, B, H]
        # state:   [1, B, H]
        return self.rnn(
            inputs,
            state,
        )
```

```python
# Concise RNN Language Model

class RNNLM(d2l.RNNLMScratch):
    rnn: RNN
    vocab_size: int

    def init_params(self) -> None:
        self.linear = nn.LazyLinear(
            self.vocab_size
        )
        
    def output_layer(
        self, 
        hiddens: torch.Tensor,
    ) -> torch.Tensor:
        
        # [T, B, H] -> [T, B, V]
        logits = self.linear(
            hiddens
        )
        
        # Cross-Entropy가 사용하는 순서로 swapping
        #
        # [T, B, V] -> [B, T, V]
        return logits.swapaxes(
            0,
            1,
        )
```

```python
# Concise RNN Shape Check

batch_size = 2
num_steps = 3
vocab_size = 4
num_hiddens = 8

rnn = RNN(
    num_inputs=vocab_size,
    num_hiddens=num_hiddens,
)

model = RNNLM(
    rnn=rnn,
    vocab_size=vocab_size,
    lr=1,
)

X = torch.tensor([
    [0, 1, 2],
    [1, 2, 3],
])

# [B, T] -> [T, B, V]
one_hot_inputs = model.one_hot(X)

hiddens, final_state = rnn(
    one_hot_inputs
)

# [B, T] -> [B, T, V]
outputs = model(X)

print(
    "Token indices:",
    tuple(X.shape),
)
print(
    "One-hot inputs:",
    tuple(one_hot_inputs.shape),
)
print(
    "RNN hiddens:",
    tuple(hiddens.shape),
)
print(
    "Final state:",
    tuple(final_state.shape),
)
print(
    "Vocabulary logits:",
    tuple(outputs.shape),
)

assert one_hot_inputs.shape == (
    num_steps,
    batch_size,
    vocab_size,
)

assert hiddens.shape == (
    num_steps,
    batch_size,
    num_hiddens,
)

assert final_state.shape == (
    1,
    batch_size,
    num_hiddens,
)

assert outputs.shape == (
    batch_size,
    num_steps,
    vocab_size,
)
```

```python
# Prediction Before Training

data = d2l.TimeMachine(
    batch_size=1024,
    num_steps=32,
)

vocab_size = len(
    data.vocab
)

rnn = RNN(
    num_inputs=vocab_size,
    num_hiddens=32,
)

model = RNNLM(
    rnn=rnn,
    vocab_size=vocab_size,
    lr=1,
)

prediction_before_training = model.predict(
    prefix="it has",
    num_preds=20,
    vocab=data.vocab,
)

print(
    prediction_before_training
)
```

```python
# High-Level RNN Training
# ==========================
# nn.RNN Forward
# -> Cross Entropy
# -> Backward
# -> Gradient Clipping
# -> Parameter Update

trainer = d2l.Trainer(
    max_epochs=100,
    gradient_clip_val=1,
    num_gpus=1,
)

trainer.fit(
    model,
    data,
)
```

```python
# Prediction After Training

prediction_after_training = model.predict(
    prefix="it has",
    num_preds=20,
    vocab=data.vocab,
    device=d2l.try_gpu(),
)

model_device = next(
    model.parameters()
).device

print(
    "Device:",
    model_device,
)
print(
    "Generated text:",
    prediction_after_training,
)
```
