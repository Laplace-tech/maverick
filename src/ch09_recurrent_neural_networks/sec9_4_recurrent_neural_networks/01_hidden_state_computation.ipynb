```python
import torch
from d2l import torch as d2l
```

```python
# Separate Matrix Multiplication

batch_size = 3
input_size = 1
hidden_size = 4

X = torch.randn(
    batch_size,
    input_size,
)

W_xh = torch.randn(
    input_size,
    hidden_size,
)

H_previous = torch.randn(
    batch_size,
    hidden_size,
)

W_hh = torch.randn(
    hidden_size,
    hidden_size,
)

separate_result = (
    X @ W_xh
    + H_previous @ W_hh
)
```

```python
print(
    "X shape:",
    tuple(X.shape),
)

print(
    "W_xh shape:",
    tuple(W_xh.shape),
)

print(
    "H_previous shape:",
    tuple(H_previous.shape),
)

print(
    "W_hh shape:",
    tuple(W_hh.shape),
)

print(
    "Result shape:",
    tuple(separate_result.shape),
)

print(
    "\nSeparate result:"
)

print(
    separate_result
)
```

```python
# Concatenated Matrix Multiplication

# X concatenates H_previous: [3, 1+4]
combined_input = torch.cat(
    (
        X,
        H_previous,
    ),
    dim=1,
)

# W_xh concatenates W_hh: [1+4, 4]
combined_weight = torch.cat(
    (
        W_xh,
        W_hh,
    ),
    dim=0,
)

concatenated_result = (
    combined_input
    @ combined_weight
)
```

```python

print(
    "Combined input shape:",
    tuple(combined_input.shape),
)

print(
    "Combined weight shape:",
    tuple(combined_weight.shape),
)

print(
    "Result shape:",
    tuple(concatenated_result.shape),
)

print(
    "Same result:",
    torch.allclose(
        separate_result,
        concatenated_result,
    ),
)


assert combined_input.shape == (
    batch_size,
    input_size + hidden_size,
)

assert combined_weight.shape == (
    input_size + hidden_size,
    hidden_size,
)

assert torch.allclose(
    separate_result,
    concatenated_result,
)
```
